```python
import pandas as pd 
```

```python
dane = pd.read_csv(
    "top10_stocks_data.csv",
    header=[0, 1],
    index_col=0
)
```

```python
dane.head()
```

```python
ceny_zamkniecia = dane['Close'].astype(float)
```

```python
stopy_zwrotu = ceny_zamkniecia.pct_change() * 100
```

```python
stopy_zwrotu
```

```python
avg_stopy_zwrotu = stopy_zwrotu.mean().sort_values(ascending=False)
avg_stopy_zwrotu
```

```python
total_return = (ceny_zamkniecia.iloc[-1] - ceny_zamkniecia.iloc[0]) / ceny_zamkniecia.iloc[0] * 100
```

```python
total_return.sort_values(ascending=False)
```

```python
#srednia kroczaca z 20 dni: 

sma_20 = ceny_zamkniecia.rolling(window=20).mean()

sma_20.tail(20)
```

```python
#srednia kroczaca z 50 dni: 

sma_50 = ceny_zamkniecia.rolling(window=50).mean()

sma_50.tail(50)
```

```python
zmiennosc_rank = stopy_zwrotu.std().sort_values(ascending=False)
zmiennosc_rank
```

```python
wolumen = dane['Volume']
wolumen.mean().sort_values(ascending=False)
```

```python
total_return.idxmax()
```

```python
total_return.idxmin()
```

```python
total_return.to_csv("total_returns.csv")
sma_20.to_csv("sma_20.csv")
sma_50.to_csv("sma_50.csv")
zmiennosc_rank.to_csv("volatility.csv")
wolumen.mean().to_csv("average_volume.csv")
stopy_zwrotu.to_csv("dane/daily_returns.csv")
```
