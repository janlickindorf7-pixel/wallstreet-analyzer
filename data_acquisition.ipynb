```python
import yfinance as yf
```

```python
import pandas as pd

url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"

headers = {
    "User-Agent": "Mozilla/5.0"
}

tabela = pd.read_html(url, storage_options=headers)
df_sp500 = tabela[0]


tickers = df_sp500['Symbol'].str.replace('.', '-', regex=False).tolist()
```

```python
kapitalizacja_spolek = []

for ticker in tickers:
    spolka = yf.Ticker(ticker)
    kapitalizacja = spolka.info["marketCap"]

    kapitalizacja_spolek.append({
        "ticker": ticker,
        "market_cap": kapitalizacja
    })
```

```python
kapitalizacja_df = pd.DataFrame(kapitalizacja_spolek)
kapitalizacja_df_sorted = kapitalizacja_df.sort_values(by='market_cap', ascending=False)
kapitalizacja_df_sorted.reset_index(drop=True, inplace=True)

top_tickers = kapitalizacja_df_sorted["ticker"].iloc[:10].tolist()

dane = yf.download(
    top_tickers,
    period="1y",
    auto_adjust=False
)

dane = dane[["Open", "High", "Low", "Close", "Volume"]]

to_csv= dane.to_csv("top10_stocks_data.csv", index=True)
```
